```python
from google.colab import files
import pandas as pd
import numpy as np
import io
import matplotlib.pyplot as plt
%matplotlib inline
uploaded = files.upload()
bank_marketing_filename = next(iter(uploaded))
b_marketing = pd.read_csv(io.StringIO(uploaded[bank_marketing_filename].decode('utf-8')))
display(b_marketing.head())
b_marketing.info()
```

```python
X = b_marketing.columns.drop("y")
Y = b_marketing['y']
```

```python
b_marketing_encoded = pd.get_dummies(b_marketing[X])
print("Total no.of predictors after encoding =  ",len(b_marketing_encoded.columns))
b_marketing_encoded.columns
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test= train_test_split(b_marketing_encoded, Y , test_size=0.20,random_state=100)
print("X_train shape = ",X_train.shape)
print("X_test shape = ",X_test.shape)
print("Y_train shape = ",Y_train.shape)
print("Y_test shape = ",Y_test.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier(random_state=1)
model.fit(X_train,Y_train)
train_pred = model.predict(X_train)
test_pred = model.predict(X_test)
```

```python
train_accuracy = model.score(X_train,Y_train)
test_accuracy = model.score(X_test,Y_test)
print("Train accuracy = ",train_accuracy)
print("Test accuracy = ",test_accuracy)
```

```python
train_precision = model.score(X_train,Y_train)
test_precision = model.score(X_test,Y_test)
print("Train Precision = ",train_precision)
print("Test Precision = ",test_precision)
```

```python
train_f1_score = model.score(X_train,Y_train)
test_f1_score = model.score(X_test,Y_test)
print("Train F1 Score = ",train_f1_score)
print("Test F1 Score = ",test_f1_score)
```

```python
model1 = DecisionTreeClassifier(min_samples_split=200,min_impurity_decrease=0.001)
model1.fit(X_train,Y_train)
print("Train accuracy = ",model1.score(X_train,Y_train))
print("Test accuracy = ",model1.score(X_test,Y_test))
```

```python
model2 = DecisionTreeClassifier(min_samples_split=1000,min_impurity_decrease=0.1)
model2.fit(X_train,Y_train)
print("Train accuracy = ",model2.score(X_train,Y_train))
print("Test accuracy = ",model2.score(X_test,Y_test))
```

```python
from sklearn.tree import export_graphviz
import graphviz
dot_data = export_graphviz(model, out_file=None,
                           feature_names=b_marketing_encoded.columns,
                           class_names=model.classes_,)
graph = graphviz.Source(dot_data)
graph
```
