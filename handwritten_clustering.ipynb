```python
from google.colab import files
import pandas as pd
import numpy as np
import io
import matplotlib.pyplot as plt
%matplotlib inline
uploaded = files.upload()
handwritten_clustering_filename= next(iter(uploaded))
h_clustering_data = np.genfromtxt(handwritten_clustering_filename, delimiter=',')
print(h_clustering_data)
```

```python
h_clustering_data = pd.read_csv(handwritten_clustering_filename)
im = h_clustering_data.iloc[0:1,:].values.reshape(28,28)
plt.imshow(im,cmap=plt.cm.gray)
```

```python
from sklearn.cluster import KMeans
model = KMeans(n_clusters=10)
model.fit(h_clustering_data)
print(np.unique(model.labels_))
```

```python
cluster1 = h_clustering_data[model.labels_==0]
cluster1_imgs = cluster1.iloc[[np.random.randint(0,cluster1.shape[0]) for i in range(0,5)]]
for i in range(0,cluster1_imgs.shape[0]):
    plt.subplot(1,5,i+1)
    img_fig = np.asarray(cluster1_imgs[i:i+1]).reshape(28,28)
    plt.imshow(img_fig,cmap=plt.cm.gray)
```

```python
# Explore images in cluster 2
cluster2 = h_clustering_data[model.labels_==1]
cluster2_imgs = cluster2.iloc[[np.random.randint(0,cluster2.shape[0]) for i in range(0,5)]]
for i in range(0,cluster2_imgs.shape[0]):
    plt.subplot(1,5,i+1)
    img_fig = np.asarray(cluster2_imgs[i:i+1]).reshape(28,28)
    plt.imshow(img_fig,cmap=plt.cm.gray)
```
