```python
from google.colab import files
import numpy as np
import pandas as pd
import io
import matplotlib.pyplot as plt
%matplotlib inline
uploaded = files.upload()
note_rfc_filename = next(iter(uploaded))

# Initial read, which puts the combined string into the 'intrest' column and NaNs in others.
n_rfc = pd.read_csv(io.StringIO(uploaded[note_rfc_filename].decode('utf-8')))

# Split the 'intrest' column into separate columns based on the comma delimiter
split_data = n_rfc['intrest'].str.split(',', expand=True)

# Assign the split parts to the correct columns and convert them to numeric types
n_rfc['intrest'] = pd.to_numeric(split_data[0])
n_rfc['amount'] = pd.to_numeric(split_data[1])
n_rfc['s'] = pd.to_numeric(split_data[2])
n_rfc['e'] = pd.to_numeric(split_data[3])
n_rfc['y'] = pd.to_numeric(split_data[4])

# Display the head of the corrected DataFrame
display(n_rfc.head())
```

```python
from re import X
from sklearn.model_selection import train_test_split
features = n_rfc.columns.drop('y')
target="y"
X=n_rfc[features]
Y=n_rfc[target]
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3,random_state=45)
```

```python
from sklearn.ensemble import RandomForestClassifier
model= RandomForestClassifier(n_estimators=20,min_samples_split=30,min_impurity_decrease=0.05)
model.fit(X_train,Y_train)
train_accuracy=model.score(X_train,Y_train)
test_accuracy=model.score(X_test,Y_test)
print("train accuracy= ",train_accuracy)
print("test accuracy= ",test_accuracy)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import AdaBoostClassifier
model = AdaBoostClassifier(n_estimators=200)
model.fit(X_train,Y_train)
train_accuracy=model.score(X_train,Y_train)
test_accuracy=model.score(X_test,Y_test)
print("decision tree train accuracy= ",train_accuracy)
print("decision tree test accuracy= ",test_accuracy)
```

```python
from sklearn.linear_model import LogisticRegression
logistic_model = LogisticRegression(random_state=45, solver='liblinear')
logistic_model.fit(X_train, Y_train)
logistic_train_accuracy = logistic_model.score(X_train, Y_train)
logistic_test_accuracy = logistic_model.score(X_test, Y_test)

print(f"Logistic Regression Train Accuracy: {logistic_train_accuracy}")
print(f"Logistic Regression Test Accuracy: {logistic_test_accuracy}")
```
